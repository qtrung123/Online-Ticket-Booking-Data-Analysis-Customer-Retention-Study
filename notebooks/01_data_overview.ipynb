```python
import pandas as pd
from pathlib import Path

BASE_DIR = Path.cwd().parent
RAW_DIR = BASE_DIR / "data" / "raw" / "movie_ticket_data" / "movie_ticket_data"

df_campaign = pd.read_csv(RAW_DIR / "campaign.csv")
df_customer = pd.read_csv(RAW_DIR / "customer.csv")
df_device = pd.read_csv(RAW_DIR / "device_detail.csv")
df_status = pd.read_csv(RAW_DIR / "status_detail.csv")
df_ticket = pd.read_csv(RAW_DIR / "ticket_history.csv")
```

# Save data

```python
from pathlib import Path

BASE_DIR = Path.cwd().parent
PROCESSED_DIR = BASE_DIR / "data" / "processed"
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

df_device_clean.to_csv(PROCESSED_DIR / "device_clean.csv", index=False)
df_campaign_clean.to_csv(PROCESSED_DIR / "campaign_clean.csv", index=False)
df_status_clean.to_csv(PROCESSED_DIR / "status_clean.csv", index=False)
df_ticket_clean.to_csv(PROCESSED_DIR / "ticket_clean.csv", index=False)

print(" All cleaned data saved!")
```

# Join tables

```python
df_fact_ticket = df_ticket_clean.copy()

df_fact_ticket = df_fact_ticket.merge(
    df_customer_clean, how='left', on='customer_id'
)

df_fact_ticket = df_fact_ticket.merge(
    df_campaign_clean, how='left', on='campaign_id'
)

df_fact_ticket = df_fact_ticket.merge(
    df_status_clean, how='left', on='status_id'
)

df_fact_ticket = df_fact_ticket.merge(
    df_device_clean, how='left', on='device_number'
)
```

```python
df_fact_ticket.isnull().sum()
```

```python
df_fact_ticket["campaign_type"] = df_fact_ticket["campaign_type"].fillna("no_campaign")
df_fact_ticket["model"] = df_fact_ticket["model"].fillna("unknown")
df_fact_ticket["platform"] = df_fact_ticket["platform"].fillna("unknown")
```

```python
df_fact_ticket["usergender"] = df_fact_ticket["usergender"].fillna("unknown")
```

```python
df_fact_ticket = df_fact_ticket.dropna(subset=["dob"])
```

```python
df_fact_ticket.isnull().sum()
```

```python
df_fact_ticket.to_csv(PROCESSED_DIR / "fact_ticket.csv", index=False)

print("fact_ticket saved")
```
